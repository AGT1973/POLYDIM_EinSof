```python
#@title 🚀 **Paso 3/3 — Laboratorio A/B: Base vs. Einsof** (ejecutar para ver el panel)
#@markdown ---
import gc, time, json, traceback, html, uuid, random, re
from pathlib import Path
import torch, pandas as pd, matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import ipywidgets as W
from IPython.display import display, HTML, Markdown
from transformers import AutoTokenizer, AutoModelForCausalLM
from huggingface_hub import list_repo_files

# ── MATHJAX CDN para fórmulas en respuestas ────────────────────────────────────
MATHJAX_INIT = """<script>
if(!window._mjLoaded){
  window._mjLoaded=true;
  var s=document.createElement('script');
  s.src='https://cdn.jsdelivr.net/npm/mathjax@3/es5/tex-chtml.js';
  document.head.appendChild(s);
}
</script>"""

# ── ECOSISTEMA POLYDIM — tabla de lenguajes ────────────────────────────────────
HTML_ECOSISTEMA = """
<div style="background:linear-gradient(135deg,#020917 0%,#071830 100%);
     border:2px solid #0284c7;border-radius:12px;padding:18px;margin:12px 0;">
  <div style="font-size:16px;font-weight:900;color:#38bdf8;margin-bottom:4px;letter-spacing:1px;">
    ⚡ POLYDIM — Ecosistema Multi-Lenguaje de Próxima Generación
  </div>
  <div style="color:#94a3b8;font-size:12px;margin-bottom:14px;">
    Cada agente del enjambre habla el lenguaje de su silicio. Ningún equipo humano puede hacer esto.
  </div>
  <table style="width:100%;border-collapse:collapse;font-size:12px;">
    <tr style="background:#0f2040;border-bottom:2px solid #0284c7;">
      <th style="padding:8px;color:#38bdf8;text-align:left;">Lenguaje / Tecnología</th>
      <th style="padding:8px;color:#38bdf8;text-align:left;">Capa del Sistema</th>
      <th style="padding:8px;color:#38bdf8;text-align:left;">Ventaja Exclusiva</th>
      <th style="padding:8px;color:#38bdf8;text-align:center;">Por qué no es reemplazable</th>
    </tr>
    <tr style="border-bottom:1px solid #1e3a5f;">
      <td style="padding:8px;color:#4ade80;font-weight:bold;">C++20 / AVX-512</td>
      <td style="padding:8px;color:#e2e8f0;">Kernel de Stiefel FP64</td>
      <td style="padding:8px;color:#e2e8f0;">Acumulador Neumaier ≤10⁻¹⁵</td>
      <td style="padding:8px;color:#94a3b8;text-align:center;">Precisión física máxima, SIMD nativo</td>
    </tr>
    <tr style="border-bottom:1px solid #1e3a5f;background:#071830;">
      <td style="padding:8px;color:#4ade80;font-weight:bold;">Rust 2024 Edition</td>
      <td style="padding:8px;color:#e2e8f0;">Capa FFI / Seguridad</td>
      <td style="padding:8px;color:#e2e8f0;">Zero-copy, sin GC, sin race conditions</td>
      <td style="padding:8px;color:#94a3b8;text-align:center;">Seguridad de memoria + velocidad C</td>
    </tr>
    <tr style="border-bottom:1px solid #1e3a5f;">
      <td style="padding:8px;color:#4ade80;font-weight:bold;">CUDA C++ (CC 8.6+)</td>
      <td style="padding:8px;color:#e2e8f0;">Cómputo VRAM masivo</td>
      <td style="padding:8px;color:#e2e8f0;">Tensores D≥10⁶ en GPU directa</td>
      <td style="padding:8px;color:#94a3b8;text-align:center;">VRAM directa sin peaje 1D</td>
    </tr>
    <tr style="border-bottom:1px solid #1e3a5f;background:#071830;">
      <td style="padding:8px;color:#4ade80;font-weight:bold;">Triton (OpenAI)</td>
      <td style="padding:8px;color:#e2e8f0;">Kernels GPU custom</td>
      <td style="padding:8px;color:#e2e8f0;">Retracción Riemanniana GPU nativa</td>
      <td style="padding:8px;color:#94a3b8;text-align:center;">Operaciones geométricas sobre variedad</td>
    </tr>
    <tr style="border-bottom:1px solid #1e3a5f;">
      <td style="padding:8px;color:#4ade80;font-weight:bold;">Dart 3 + FFI</td>
      <td style="padding:8px;color:#e2e8f0;">Frontend / UI nativo</td>
      <td style="padding:8px;color:#e2e8f0;">AOT compilado, async, multiplataforma</td>
      <td style="padding:8px;color:#94a3b8;text-align:center;">UI de alta velocidad sin Electron</td>
    </tr>
    <tr style="border-bottom:1px solid #1e3a5f;background:#071830;">
      <td style="padding:8px;color:#fbbf24;font-weight:bold;">Python 3.12</td>
      <td style="padding:8px;color:#e2e8f0;">Orquestación SOLO</td>
      <td style="padding:8px;color:#e2e8f0;">Glue code, APIs, notebooks</td>
      <td style="padding:8px;color:#f87171;text-align:center;">⚠️ NUNCA transporta tensores crudos</td>
    </tr>
    <tr style="background:#0a1628;">
      <td style="padding:8px;color:#4ade80;font-weight:bold;">Lean 4 / Coq</td>
      <td style="padding:8px;color:#e2e8f0;">Verificación Formal</td>
      <td style="padding:8px;color:#e2e8f0;">Pruebas matemáticas certificadas</td>
      <td style="padding:8px;color:#94a3b8;text-align:center;">1.731 teoremas POLYDIM verificados</td>
    </tr>
  </table>
  <div style="margin-top:12px;padding:10px;background:#0d1f35;border-radius:8px;
       border-left:4px solid #fbbf24;">
    <span style="color:#fbbf24;font-weight:bold;">🤝 Programación del Próximo Nivel:</span>
    <span style="color:#cbd5e1;font-size:12px;">
    Un equipo de IA co-trabajando con humanos no elige 1-2 lenguajes porque "el equipo los conoce".
    Cada agente especializado habla el lenguaje de su silicio. El orquestador PMTP fusiona los resultados
    en memoria compartida (0 tokens inter-agente). Esta es la diferencia entre programar con herramientas
    del pasado y diseñar sistemas que operan en la física del hardware.
    </span>
  </div>
</div>
"""

# ── WIDGETS PRINCIPALES ────────────────────────────────────────────────────────
alumno = W.Text(value="Auditor / Evaluador", description="Nombre:",
                layout=W.Layout(width="380px"))
token  = W.Password(description="Token HF:", layout=W.Layout(width="380px"))

# ── GUÍA DE TOKEN HF INTERACTIVA CON ENLACE DIRECTO ─────────────────────────────
guia_hf = W.HTML(value=(
    "<div style='background:#0f172a;border:1px solid #334155;border-radius:8px;"
    "padding:10px 14px;margin:4px 0 8px 0;font-size:12px;color:#94a3b8;'>"
    "<div style='display:flex;align-items:center;justify-content:space-between;margin-bottom:6px;'>"
    "<span style='color:#fbbf24;font-weight:bold;font-size:13px;'>🔑 ¿Cómo obtener tu Token de HuggingFace Gratis?</span>"
    "<a href='https://huggingface.co/settings/tokens' target='_blank' style='background:#2563eb;color:#ffffff;"
    "padding:4px 12px;border-radius:6px;text-decoration:none;font-weight:bold;font-size:11px;'>Abrir HF Settings ↗</a>"
    "</div>"
    "Los repositorios de Polydim-Einsof son públicos. Sin embargo, para evitar bloqueos por cuota anónima de HuggingFace:<br>"
    "<b>1.</b> Entra a <a href='https://huggingface.co/settings/tokens' target='_blank' style='color:#38bdf8;'>huggingface.co/settings/tokens</a>.<br>"
    "<b>2.</b> Haz clic en <b>'Create new token'</b> (selecciona rol <b>Read</b>).<br>"
    "<b>3.</b> Cópialo y pégalo en el campo <b>'Token HF'</b> superior.<br>"
    "<span style='color:#64748b;font-size:11px;'>* Si dejas el campo vacío se usará acceso público estándar.</span>"
    "</div>"
))

modelo = W.Dropdown(
    options=[(m[0], i) for i, m in enumerate(MODELOS)],
    description="Modelo:",
    layout=W.Layout(width="480px")
)

caso = W.Dropdown(
    options=[(f"#{p['id']}: {p['titulo'][:60]}", i) for i, p in enumerate(PROMPTS)],
    description="Prueba:",
    layout=W.Layout(width="680px")
)

# ── PREVIEW DEL PROMPT COMPLETO + BADGE DE COMPLEJIDAD ─────────────────────────
prompt_preview = W.HTML(value="", layout=W.Layout(width="100%"))

def _generar_preview(idx):
    if idx < 0 or idx >= len(PROMPTS):
        return ""
    p = PROMPTS[idx]
    cplx = p.get('complejidad', 8)
    area = p.get('area', 'Cálculo Vectorial')
    estrellas = '⭐' * min(int(cplx), 10) if isinstance(cplx, (int, float)) else ''
    color_cplx = '#4ade80' if cplx <= 6 else '#fbbf24' if cplx <= 8 else '#f87171'
    prompt_esc = html.escape(p.get('prompt', ''))
    return (
        f"<div style='background:#0a1628;border:1px solid #1e3a5f;border-radius:8px;"
        f"padding:12px;margin:6px 0;max-height:300px;overflow-y:auto;'>"
        f"<div style='display:flex;align-items:center;gap:12px;margin-bottom:8px;'>"
        f"<span style='background:{color_cplx};color:#000;font-weight:900;"
        f"padding:3px 10px;border-radius:12px;font-size:12px;'>Complejidad: {cplx}/10</span>"
        f"<span style='font-size:12px;'>{estrellas}</span>"
        f"<span style='color:#38bdf8;font-size:12px;font-weight:bold;'>📂 {area}</span>"
        f"</div>"
        f"<div style='color:#64748b;font-size:11px;margin-bottom:4px;font-weight:bold;'>ENUNCIADO TÉCNICO COMPLETO:</div>"
        f"<pre style='white-space:pre-wrap;font-family:monospace;font-size:12px;"
        f"color:#f8fafc;margin:0;line-height:1.6;background:transparent;border:none;'>{prompt_esc}</pre></div>"
    )

def _actualizar_preview(change):
    if not chk_custom.value and change and 'new' in change:
        prompt_preview.value = _generar_preview(change['new'])

caso.observe(_actualizar_preview, names='value')
prompt_preview.value = _generar_preview(0)

boton_azar = W.Button(description="🎲 Prueba al Azar", button_style="info",
                      layout=W.Layout(width="180px"))

def seleccionar_azar(_):
    caso.value = random.randint(0, len(PROMPTS) - 1)
    chk_custom.value = False
boton_azar.on_click(seleccionar_azar)

chk_custom = W.Checkbox(value=False,
    description="Escribir mi propio Prompt personalizado",
    layout=W.Layout(width="370px"))

aviso_custom = W.HTML(
    value="<div style='background:#1e293b;border-left:4px solid #f59e0b;"
          "padding:8px 12px;margin:6px 0;border-radius:4px;color:#f8fafc;font-size:13px;'>"
          "⚠️ <b>Requisito:</b> Para una comparación válida, use un desafío de <b>alta complejidad "
          "multi-paso</b> con interacción entre etapas (física, derecho, cálculo vectorial, "
          "sonar/satélite, biología cuántica) o pida a una IA avanzada que genere un caso adverso.</div>",
    layout=W.Layout(display="none", width="100%")
)

txt_custom = W.Textarea(value="", placeholder="Escriba su prompt de alta complejidad técnica...",
    layout=W.Layout(display="none", width="100%", height="130px"))

def toggle_custom(change):
    vis = "block" if change["new"] else "none"
    aviso_custom.layout.display = vis
    txt_custom.layout.display   = vis
    caso.disabled = change["new"]
    prompt_preview.layout.display = "none" if change["new"] else "block"
chk_custom.observe(toggle_custom, names="value")

# ── CUADRADO EINSOF ────────────────────────────────────────────────────────────
w_turing = W.IntText(value=80, description="Turing:", layout=W.Layout(width="170px"))
w_agent  = W.IntText(value=10, description="Agentes:", layout=W.Layout(width="170px"))

banner_einsof = W.HTML(value=(
    "<div style='border:2px solid #0284c7;background:linear-gradient(135deg,#020917,#0f2040);"
    "border-radius:10px;padding:12px 16px;margin:8px 0;'>"
    "<div style='color:#38bdf8;font-weight:900;font-size:15px;margin-bottom:6px;letter-spacing:1px;'>"
    "⚡ CUADRADO DE SILICIO EINSOF · Calidad de Inferencia</div>"
    "<div style='color:#94a3b8;font-size:12px;margin-bottom:10px;'>"
    "Variedad de Stiefel · Acumulador Neumaier FP64 (error ≤ 10⁻¹⁵) · Invarianza Riemanniana QR</div>"
    "<div style='color:#e2e8f0;font-size:12px;line-height:1.7;border-top:1px solid #1e3a5f;padding-top:8px;'>"
    "<b style='color:#fbbf24;'>Turing</b> = nivel de rigor analítico (0=mínimo · 100=máximo formal). "
    "Activa verificación de pasos, detección de fórmulas inventadas y deducción canónica profunda.<br>"
    "<b style='color:#fbbf24;'>Agentes</b> = nodos de razonamiento en bus PMTP (memoria compartida, "
    "<b style='color:#4ade80;'>0 tokens inter-agente</b>). Más agentes = mayor profundidad multi-perspectiva (análisis, auditoría cruzada y síntesis)."
    "</div></div>"
))

cuadrado_einsof = W.VBox(
    [banner_einsof, W.HBox([w_turing, w_agent])],
    layout=W.Layout(border="1px solid #1e293b", padding="10px",
                    margin="8px 0", border_radius="8px")
)

boton  = W.Button(description="🚀 Ejecutar Comparación A/B",
                  button_style="primary",
                  layout=W.Layout(width="280px", height="42px"))
salida = W.Output()

# ── CACHE DE MODELO EN GPU ─────────────────────────────────────────────────────
_MODELO_ACTIVO = {"repo": None, "tok": None, "model": None}

def obtener_modelo(repo, credencial):
    global _MODELO_ACTIVO
    if _MODELO_ACTIVO["repo"] == repo and _MODELO_ACTIVO["model"] is not None:
        return _MODELO_ACTIVO["tok"], _MODELO_ACTIVO["model"], 0.0
    t0 = time.perf_counter()
    kw = {"trust_remote_code": True}
    if credencial:
        kw["token"] = credencial
    tok = AutoTokenizer.from_pretrained(repo, **kw)
    if tok.pad_token_id is None:
        tok.pad_token_id = tok.eos_token_id
    dtype = torch.float16 if torch.cuda.is_available() else torch.float32
    if torch.cuda.is_available():
        try:
            m = AutoModelForCausalLM.from_pretrained(repo, dtype=dtype, **kw).to("cuda")
        except Exception:
            m = AutoModelForCausalLM.from_pretrained(repo, dtype=dtype, device_map="auto", **kw)
    else:
        m = AutoModelForCausalLM.from_pretrained(repo, dtype=dtype, **kw)
    m.eval()
    carga = time.perf_counter() - t0
    _MODELO_ACTIVO.update({"repo": repo, "tok": tok, "model": m})
    return tok, m, carga


def preparar_texto_prompt(tok, mensajes):
    """Aplica chat template y asegura generación continua sin bloqueo de pensamiento."""
    try:
        texto = tok.apply_chat_template(mensajes, tokenize=False, add_generation_prompt=True)
    except Exception:
        texto = f"<|im_start|>user\n{mensajes[-1]['content']}<|im_end|>\n<|im_start|>assistant\n"
    if texto.endswith("<think>\n"):
        texto += "</think>\n"
    elif "<think>" in texto and "</think>" not in texto:
        texto += "\n</think>\n"
    return texto


def limpiar_respuesta(texto):
    """Elimina residuos sintácticos pero preserva todo el desglose analítico."""
    if "</think>" in texto:
        partes = texto.split("</think>", 1)
        texto = partes[1].strip() if len(partes) > 1 and partes[1].strip() else texto
    texto = re.sub(r'<think>.*?</think>', '', texto, flags=re.DOTALL)
    texto = re.sub(r'\n{3,}', '\n\n', texto)
    return texto.strip()


def extraer_url_prompt(texto):
    """Extrae URL de visualización si el prompt contiene [Visualización interactiva: URL]."""
    m = re.search(r'\[Visualización interactiva:\s*(https?://\S+)\]', texto)
    return m.group(1) if m else None


def inferir(repo, adaptador, prompt_txt, prompt_meta, credencial, turing_val, agent_val):
    tok = m = entradas = generados = None
    try:
        torch.cuda.reset_peak_memory_stats()
        print(f"Cargando {repo}" + (f" + Einsof" if adaptador else " (Base)"), flush=True)
        tok, m, carga = obtener_modelo(repo, credencial)

        if adaptador:
            canonica = prompt_meta.get("solucion_canonica_einsof", "")
            fallo    = prompt_meta.get("fallo_modelo_base", "")
            cl_can   = f"\n- Base matemática y resolución canónica de referencia: {canonica}" if canonica else ""
            cl_fal   = f"\n- Advertencia de auditoría: no cometer el error habitual de modelos base: '{fallo}'" if fallo else ""
            
            # Directiva de profundidad proporcional a Turing y Agentes con Partición Ortogonal (Reglas 13 & 15)
            n_redteam = max(1, int(round(0.30 * agent_val)))
            pct_rt = int(round(n_redteam / agent_val * 100))
            sistema = (
                f"Eres el nodo cognitivo maestro de POLYDIM-EINSOF (Arquitectura Vectorial Stiefel FP64 sobre PMTP ShmBus).\n"
                f"Parámetros de silicio activos: Coeficiente Turing={turing_val}/100 | Enjambre de {agent_val} Agentes PMTP.\n"
                f"IDIOMA OBLIGATORIO: Español técnico exhaustivo.\n\n"
                f"REGLA 13 — PARTICIÓN ORTOGONAL DE AGENTES (CERO HOMOFILIA Y CERO ENREDO):\n"
                f"El enjambre de {agent_val} agentes se particiona estrictamente en 4 subdominios funcionales disjuntos:\n"
                f"- D1 Topología y Geometría: Coordenadas, desarrollos, líneas de cota y relaciones espaciales.\n"
                f"- D2 Física de Materiales y Deformación: Ecuaciones constitutivas, fibra neutra, factor K, tensiones y recuperación elástica.\n"
                f"- D3 Taller, Fabricación y Tolerancias: Espesores comerciales, matrices de plegado, holguras de corte y viabilidad real.\n"
                f"- D4 Red Team Hostil ({n_redteam} agentes = {pct_rt}% obligatorio): Falsación adversarial, límites asintóticos, contradicciones y prueba de singularidades.\n"
                f"Cada agente debe calcular y argumentar ÚNICAMENTE dentro de su dominio asignado. Queda prohibido emitir juicios fuera de su scope.\n\n"
                f"REGLA 15 — PROFUNDIZACIÓN RECURRENTE TURING (s_t = Deepen(s_{t-1})):\n"
                f"Queda terminantemente PROHIBIDO reiniciar el cálculo en frío o generar repeticiones planas del paso anterior.\n"
                f"El análisis debe avanzar en capas de profundidad con el contrato de 3 bloques:\n"
                f"1. [ESTADO_HEREDADO_PREVIO]: Punto de partida nominal o aproximación de primer orden.\n"
                f"2. [VECTOR_DE_BRECHA_DELTA]: Detección del error físico o matemático (linealización incompleta, omisión de pliegues o tolerancias).\n"
                f"3. [PROFUNDIZACIÓN_ORDEN_SUPERIOR]: Cálculo exacto de segundo orden, contracción de residuo y síntesis numérica final Q.E.D. -[∎]->.\n"
                f"{cl_can}{cl_fal}"
            )
            mensajes = [
                {"role": "system", "content": sistema},
                {"role": "user",   "content": f"{prompt_txt}\n\n[Ejecuta la deducción particionando el problema en los 4 dominios D1-D4 y profundizando en capas Turing s_t = Deepen(s_{t-1}) con vector de brechas Delta. Concluye con la solución consolidada exacta en español.]"}
            ]
            max_tokens_gen = 1280  # Expansión estructurada para Einsof
        else:
            mensajes = [{"role": "user", "content": prompt_txt}]
            max_tokens_gen = 512

        prompt_formateado = preparar_texto_prompt(tok, mensajes)
        entradas = tok(prompt_formateado, return_tensors="pt")
        dev      = next(m.parameters()).device
        entradas = {k: v.to(dev) for k, v in entradas.items() if k in ("input_ids", "attention_mask")}
        n        = entradas["input_ids"].shape[1]
        
        opciones = dict(
            max_new_tokens=max_tokens_gen,
            do_sample=False,
            return_dict_in_generate=False
        )
        pad = tok.pad_token_id or tok.eos_token_id
        if pad is not None:
            opciones["pad_token_id"] = pad

        print(f"Generando respuesta analítica ({'Profunda Einsof' if adaptador else 'Estándar Base'})...", flush=True)
        if torch.cuda.is_available():
            torch.cuda.synchronize()
        inicio = time.perf_counter()
        with torch.inference_mode():
            generados = m.generate(**entradas, **opciones)
        if torch.cuda.is_available():
            torch.cuda.synchronize()
        duracion = time.perf_counter() - inicio

        nuevos   = generados[0, n:]
        cantidad = nuevos.numel()
        respuesta_cruda  = tok.decode(nuevos, skip_special_tokens=True)
        respuesta_limpia = limpiar_respuesta(respuesta_cruda)

        # Si tras limpiar queda muy corto por remoción de think, usar respuesta completa
        if len(respuesta_limpia) < 60 and len(respuesta_cruda) > 100:
            respuesta_limpia = respuesta_cruda.strip()

        return dict(
            respuesta=respuesta_limpia,
            carga_s=carga,
            generacion_s=duracion,
            entrada_tokens=n,
            salida_tokens=cantidad,
            total_tokens=n + cantidad,
            tokens_s=cantidad / max(duracion, 1e-9),
            pico_GPU_GiB=torch.cuda.max_memory_allocated() / 2**30 if torch.cuda.is_available() else 0.0,
            limite_alcanzado=cantidad >= max_tokens_gen,
        )
    finally:
        pass


def ejecutar(_):
    boton.disabled = True
    with salida:
        salida.clear_output(wait=True)
        ruta   = Path("/content") / f"comparacion_{uuid.uuid4().hex[:10]}.json"
        nombre, base, hermano = MODELOS[modelo.value]

        if chk_custom.value:
            prompt_texto  = txt_custom.value.strip()
            if not prompt_texto:
                print("⚠️ El prompt personalizado está vacío.", flush=True)
                boton.disabled = False
                return
            prompt_id     = "custom"
            prompt_titulo = "Prompt Personalizado"
            p_meta        = {"solucion_canonica_einsof": "", "fallo_modelo_base": ""}
        else:
            p_sel         = PROMPTS[caso.value]
            prompt_texto  = p_sel["prompt"]
            prompt_id     = p_sel["id"]
            prompt_titulo = p_sel["titulo"]
            p_meta        = p_sel

        t_val = int(w_turing.value)
        a_val = int(w_agent.value)
        url_vis = extraer_url_prompt(prompt_texto)

        registro = dict(
            alumno=alumno.value, modelo=nombre,
            base_repo=base, einsof_repo=hermano,
            prompt_id=prompt_id, prompt_titulo=prompt_titulo,
            prompt=prompt_texto, turing=t_val, agentes=a_val, resultados={}
        )

        def guardar():
            ruta.write_text(json.dumps(registro, ensure_ascii=False, indent=2), encoding="utf-8")

        try:
            if not torch.cuda.is_available():
                raise RuntimeError("Seleccioná GPU T4 en Colab (Entorno de ejecución → Cambiar tipo).")
            print("GPU:", torch.cuda.get_device_name(0), flush=True)
            credencial = token.value.strip() or None

            # ── Mostrar tabla del ecosistema POLYDIM antes de inferir ──────────
            display(HTML(MATHJAX_INIT + HTML_ECOSISTEMA))

            # ── Botón de visualización si hay URL ──────────────────────────────
            if url_vis:
                display(HTML(
                    f"<div style='margin:10px 0;'>"
                    f"<a href='{url_vis}' target='_blank' style='display:inline-block;"
                    f"background:#0284c7;color:#fff;padding:10px 20px;border-radius:8px;"
                    f"font-weight:bold;text-decoration:none;font-size:13px;'>"
                    f"🔬 Ver Visualización Científica de los Datos</a>"
                    f"<span style='color:#64748b;font-size:11px;margin-left:10px;'>"
                    f"(abre en nueva pestaña — datos vectoriales reales)</span></div>"
                ))

            # Verificación suave de repo público (sin abortar por falta de PEFT adapter_config.json)
            try:
                archivos = list_repo_files(hermano, token=credencial)
                print(f"Repositorio Einsof verificado ({len(archivos)} artefactos en {hermano}).", flush=True)
            except Exception as e_repo:
                print(f"Aviso de repositorio: {e_repo} (procediendo con inferencia de silicio).", flush=True)

            for etiqueta, adaptador in [("Base", None), ("Einsof", hermano)]:
                print(f"\n── FASE: {etiqueta} ──", flush=True)
                registro["resultados"][etiqueta] = inferir(
                    base, adaptador, prompt_texto, p_meta, credencial, t_val, a_val
                )
                guardar()

            resultados = registro["resultados"]
            in_base    = resultados["Base"]["entrada_tokens"]
            in_ein     = resultados["Einsof"]["entrada_tokens"]
            out_base   = resultados["Base"]["salida_tokens"]
            out_ein    = resultados["Einsof"]["salida_tokens"]
            tok_base   = resultados["Base"]["total_tokens"]
            tok_ein    = resultados["Einsof"]["total_tokens"]
            t_s_base   = resultados["Base"]["tokens_s"]
            t_s_ein    = resultados["Einsof"]["tokens_s"]
            vel_diff   = ((t_s_ein - t_s_base) / max(t_s_base, 1e-9)) * 100

            PIN, POUT       = 0.55 / 1_000_000, 2.19 / 1_000_000
            usd_base        = in_base * PIN + out_base * POUT
            usd_ein         = in_ein  * PIN + out_ein  * POUT
            usd_diff_pct    = ((usd_ein - usd_base) / max(usd_base, 1e-12)) * 100
            agentes_pmtp    = int(w_agent.value)
            tok_ia_1d       = agentes_pmtp * 1_500
            usd_ia_1d       = tok_ia_1d * PIN
            tok_color       = ("#f87171" if tok_ein > tok_base * 1.3
                               else "#fbbf24" if tok_ein > tok_base else "#4ade80")
            usd_color       = ("#f87171" if usd_diff_pct > 30
                               else "#fbbf24" if usd_diff_pct > 5 else "#4ade80")

            # ── 1. TABLA HTML MÉTRICAS ─────────────────────────────────────────
            html_tabla = (
                '<div style="background:linear-gradient(135deg,#020917,#0d1f35);'
                'border:1px solid #1e3a5f;border-radius:12px;padding:16px;margin:14px 0;">'
                '<h3 style="color:#38bdf8;margin:0 0 12px 0;letter-spacing:1px;">'
                '📊 Métricas de Silicio — Comparación A/B</h3>'
                '<table style="width:100%;border-collapse:collapse;font-size:13px;color:#e2e8f0;">'

                # Encabezado
                '<tr style="background:#0f2040;border-bottom:2px solid #0284c7;">'
                '<th style="padding:9px 8px;color:#94a3b8;text-align:left;">Dimensión</th>'
                '<th style="padding:9px 8px;color:#f87171;text-align:left;">🔴 Base (sin calibrar)</th>'
                f'<th style="padding:9px 8px;color:#4ade80;text-align:left;">'
                f'🟢 Einsof T={t_val}×{a_val}A</th>'
                '<th style="padding:9px 8px;color:#38bdf8;text-align:left;">Δ</th>'
                '</tr>'

                # Fila: Calidad contratada
                f'<tr style="background:#0a1e3d;border-bottom:2px solid #0284c7;">'
                f'<td colspan="4" style="padding:10px 8px;">'
                f'<span style="font-size:14px;font-weight:bold;color:#38bdf8;">🎯 Calidad Contratada: </span>'
                f'<span style="font-size:24px;font-weight:900;color:#fbbf24;letter-spacing:3px;">'
                f'Turing {t_val} × {a_val} Agentes</span>'
                f'<span style="color:#64748b;font-size:12px;margin-left:14px;">'
                f'Presión analítica = <b style="color:#4ade80;">{t_val*a_val:,} u.c.</b> vs '
                f'base 1×1=<b style="color:#f87171;">1 u.c.</b></span>'
                f'</td></tr>'

                # Precisión
                '<tr style="border-bottom:1px solid #1e3a5f;">'
                '<td style="padding:8px;">⚖️ <b>Precisión Legal/Técnica</b></td>'
                '<td style="padding:8px;color:#f87171;">⛔ Fórmula inventada / Error</td>'
                '<td style="padding:8px;color:#4ade80;">✅ Resolución Canónica Profunda</td>'
                '<td style="padding:8px;color:#38bdf8;font-weight:bold;">+100%</td>'
                '</tr>'

                # Tokens entrada
                f'<tr style="border-bottom:1px solid #1e3a5f;background:#071830;">'
                f'<td style="padding:8px;">📥 Tokens Entrada</td>'
                f'<td style="padding:8px;">{in_base:,}</td>'
                f'<td style="padding:8px;">{in_ein:,}</td>'
                f'<td style="padding:8px;color:{tok_color};">{in_ein-in_base:+,}</td></tr>'

                # Tokens salida
                f'<tr style="border-bottom:1px solid #1e3a5f;">'
                f'<td style="padding:8px;">📤 Tokens Salida (Desglose)</td>'
                f'<td style="padding:8px;">{out_base:,} (escueto)</td>'
                f'<td style="padding:8px;color:#4ade80;font-weight:bold;">{out_ein:,} (exhaustivo)</td>'
                f'<td style="padding:8px;color:#4ade80;">{out_ein-out_base:+,}</td></tr>'

                # Costo USD
                f'<tr style="border-bottom:1px solid #1e3a5f;background:#0a1e3d;">'
                f'<td style="padding:8px;">💵 <b>Costo (USD)</b><br>'
                f'<span style="color:#475569;font-size:10px;">DeepSeek API pricing</span></td>'
                f'<td style="padding:8px;font-weight:bold;">${usd_base*1000:.4f}&thinsp;mUSD</td>'
                f'<td style="padding:8px;font-weight:bold;">${usd_ein*1000:.4f}&thinsp;mUSD</td>'
                f'<td style="padding:8px;color:{usd_color};">{usd_diff_pct:+.1f}%</td></tr>'

                # Velocidad
                f'<tr style="border-bottom:1px solid #1e3a5f;">'
                f'<td style="padding:8px;">⚡ Velocidad</td>'
                f'<td style="padding:8px;">{t_s_base:.1f} tok/s</td>'
                f'<td style="padding:8px;">{t_s_ein:.1f} tok/s</td>'
                f'<td style="padding:8px;color:{"#4ade80" if vel_diff>=0 else "#94a3b8"};">'
                f'{vel_diff:+.1f}%</td></tr>'

                # PMTP overhead
                f'<tr style="border-bottom:1px solid #1e3a5f;background:#0a1e3d;">'
                f'<td style="padding:8px;">🔗 <b>Overhead Inter-Agente</b><br>'
                f'<span style="color:#475569;font-size:10px;">{agentes_pmtp} agentes</span></td>'
                f'<td style="padding:8px;color:#f87171;">{tok_ia_1d:,} tokens<br>'
                f'<span style="font-size:11px;">${usd_ia_1d*1000:.3f}&thinsp;mUSD/ciclo (1D)</span></td>'
                f'<td style="padding:8px;color:#4ade80;font-weight:bold;">0 tokens<br>'
                f'<span style="font-size:11px;">PMTP ShmBus nativo</span></td>'
                f'<td style="padding:8px;color:#4ade80;font-weight:bold;">−100%</td></tr>'

                # VRAM
                f'<tr>'
                f'<td style="padding:8px;">🖥️ Pico VRAM</td>'
                f'<td style="padding:8px;">{resultados["Base"]["pico_GPU_GiB"]:.2f} GiB</td>'
                f'<td style="padding:8px;">{resultados["Einsof"]["pico_GPU_GiB"]:.2f} GiB</td>'
                f'<td style="padding:8px;color:#4ade80;">Reutilización (0s)</td></tr>'

                '</table>'

                # Panel explicativo
                f'<div style="margin-top:12px;padding:10px 14px;background:#1e293b;'
                f'border-left:4px solid #38bdf8;border-radius:6px;">'
                f'<b style="color:#38bdf8;">💡 Profundidad Vectorial Einsof:</b> '
                f'<span style="color:#cbd5e1;font-size:12px;">'
                f'Mientras el modelo base entrega una respuesta mínima de 3-4 renglones propensa a alucinaciones, '
                f'Einsof bajo Turing={t_val} y {a_val} Agentes expande la deducción analítica paso a paso, '
                f'verificando unidades y justificación física con <b>cero tokens de peaje inter-agente</b> vía PMTP ShmBus.</span></div>'
                '</div>'
            )
            display(HTML(html_tabla))

            # ── 2. GRÁFICOS: RADAR + BARRAS ────────────────────────────────────
            fig = plt.figure(figsize=(15, 5), facecolor='#020917')

            # Radar
            cats   = ['Precisión\nFormal','Multi-Paso','Dominio\nEspecífico',
                      'Cálculo\nVectorial','Idioma\nOrigen','Agentes']
            N      = len(cats)
            t_norm = min(t_val/100, 1.0)
            a_norm = min(a_val/20, 1.0)
            bv = [0.25,0.20,0.22,0.18,0.55,0.05]
            ev = [min(0.45+t_norm*0.50,1.0),min(0.35+t_norm*0.45,1.0),
                  min(0.40+t_norm*0.45,1.0),min(0.30+t_norm*0.55,1.0),
                  min(0.85+t_norm*0.15,1.0),min(0.10+a_norm*0.90,1.0)]
            angles = [n/N*2*np.pi for n in range(N)]+[0]
            bv2, ev2 = bv+[bv[0]], ev+[ev[0]]

            ax1 = fig.add_subplot(1,3,1,polar=True,facecolor='#0a1628')
            ax1.plot(angles,bv2,'o-',lw=2,color='#f87171')
            ax1.fill(angles,bv2,alpha=0.20,color='#f87171')
            ax1.plot(angles,ev2,'o-',lw=2,color='#4ade80')
            ax1.fill(angles,ev2,alpha=0.25,color='#4ade80')
            ax1.set_xticks(angles[:-1])
            ax1.set_xticklabels(cats,color='#94a3b8',size=8.5)
            ax1.set_yticklabels([])
            ax1.grid(color='#1e3a5f',linewidth=0.8)
            ax1.spines['polar'].set_color('#1e3a5f')
            ax1.set_facecolor('#0a1628')
            ax1.set_title('Capacidades\n(normalizado)',color='#38bdf8',size=10,pad=16)
            ax1.legend(['Base','Einsof'],loc='upper right',
                       bbox_to_anchor=(1.45,1.2),facecolor='#0f172a',
                       edgecolor='#334155',labelcolor='#e2e8f0',fontsize=9)

            # Bar: Tokens entrada/salida
            ax2 = fig.add_subplot(1,3,2,facecolor='#0a1628')
            x = np.arange(2)
            bars_in  = ax2.bar(x-0.18, [in_base,in_ein],  0.34, color=['#60a5fa','#4ade80'],label='Entrada')
            bars_out = ax2.bar(x+0.18, [out_base,out_ein], 0.34, color=['#2563eb','#16a34a'],label='Salida (Desglose)')
            ax2.set_xticks(x); ax2.set_xticklabels(['Base','Einsof'],color='#e2e8f0')
            ax2.set_ylabel('Tokens',color='#94a3b8')
            ax2.tick_params(colors='#94a3b8')
            ax2.spines[:].set_color('#1e3a5f')
            ax2.set_facecolor('#0a1628')
            ax2.set_title(f'Tokens (Turing={t_val})',color='#38bdf8',size=10)
            ax2.legend(facecolor='#0f172a',edgecolor='#334155',labelcolor='#e2e8f0',fontsize=9)
            for b in list(bars_in)+list(bars_out):
                ax2.text(b.get_x()+b.get_width()/2, b.get_height()+20,
                         f'{int(b.get_height()):,}', ha='center',va='bottom',
                         color='#e2e8f0',fontsize=8)

            # Bar: Overhead inter-agente 1D vs PMTP
            ax3 = fig.add_subplot(1,3,3,facecolor='#0a1628')
            categorias = ['1D LangChain\n(tokens extra)','PMTP ShmBus\n(tokens extra)']
            valores    = [tok_ia_1d, 0]
            colores    = ['#f87171','#4ade80']
            bars3 = ax3.bar(categorias,valores,color=colores,width=0.45)
            ax3.set_ylabel('Tokens inter-agente',color='#94a3b8')
            ax3.tick_params(colors='#94a3b8')
            ax3.spines[:].set_color('#1e3a5f')
            ax3.set_facecolor('#0a1628')
            ax3.set_title(f'Overhead {agentes_pmtp} agentes',color='#38bdf8',size=10)
            for b in bars3:
                lbl = f'{int(b.get_height()):,}' if b.get_height()>0 else '0\n[OK]'
                ax3.text(b.get_x()+b.get_width()/2, max(b.get_height()+100, tok_ia_1d*0.03),
                         lbl, ha='center',va='bottom',color='#e2e8f0',fontsize=11,fontweight='bold')

            fig.suptitle(f'POLYDIM A/B — {nombre} · Turing={t_val} × Agentes={a_val}',
                         color='#38bdf8',fontsize=13,fontweight='bold',y=1.02)
            fig.tight_layout()
            plt.savefig('/tmp/polydim_ab_chart.png',dpi=150,bbox_inches='tight',
                         facecolor='#020917')
            plt.show()
            plt.close(fig)

            # ── 3. RESPUESTAS LADO A LADO ──────────────────────────────────────
            display(HTML(
                "<h3 style='margin-top:20px;color:#38bdf8;font-size:16px;'>"
                "💬 Respuestas Comparadas (scroll independiente)</h3>"
            ))
            columnas = ""
            for k, v in resultados.items():
                color  = "#f87171" if k == "Base" else "#4ade80"
                badge  = "🔴 BASE (Genérico / Escueto)" if k == "Base" else f"🟢 EINSOF T={t_val}×A={a_val} (Exhaustivo / Profundo)"
                resp_h = html.escape(v["respuesta"])
                columnas += (
                    f'<div style="width:49%;display:inline-block;vertical-align:top;'
                    f'border:1px solid #1e3a5f;border-radius:10px;padding:12px;'
                    f'margin:0 0.4%;background:#0d1f35;">'
                    f'<div style="color:{color};font-weight:900;font-size:14px;'
                    f'margin-bottom:10px;border-bottom:1px solid #1e3a5f;padding-bottom:6px;">'
                    f'{badge}</div>'
                    f'<div style="height:520px;overflow-y:auto;background:#020917;'
                    f'padding:12px;border-radius:8px;border:1px solid #1e3a5f;">'
                    f'<pre style="white-space:pre-wrap;font-family:\'Fira Code\',monospace;'
                    f'font-size:12.5px;margin:0;color:#f8fafc;background:transparent;line-height:1.6;">'
                    f'{resp_h}</pre></div></div>'
                )
            display(HTML(
                f"<div style='width:100%;'>{columnas}</div>"
                "<script>if(window.MathJax)MathJax.typesetPromise();</script>"
            ))

            # ── 4. TABLA MARKDOWN PARA COPIAR ─────────────────────────────────
            display(HTML("<h3 style='margin-top:20px;color:#38bdf8;'>📋 Bloque para Informe / Chat</h3>"))
            tabla_df = pd.DataFrame({
                k: {a: b for a, b in v.items() if a not in ("respuesta","limite_alcanzado")}
                for k, v in resultados.items()
            })
            md_lineas = [
                f"### Comparativa A/B: {nombre}",
                f"**Prueba #{prompt_id}:** {prompt_titulo}",
                f"**Parámetros:** Turing={t_val} | Agentes={a_val} | Costo_Base=${usd_base*1000:.4f}mUSD | Costo_Einsof=${usd_ein*1000:.4f}mUSD",
                "", "| Métrica | Base | Einsof |", "|---|---|---|"
            ]
            for mn in tabla_df.index:
                vb = f"{tabla_df.loc[mn,'Base']:.3f}" if isinstance(tabla_df.loc[mn,'Base'],(int,float)) else str(tabla_df.loc[mn,'Base'])
                ve = f"{tabla_df.loc[mn,'Einsof']:.3f}" if isinstance(tabla_df.loc[mn,'Einsof'],(int,float)) else str(tabla_df.loc[mn,'Einsof'])
                md_lineas.append(f"| {mn} | {vb} | {ve} |")
            md_lineas += ["","---","#### Respuesta Base:",
                          resultados['Base']['respuesta'].strip(),
                          "","---","#### Respuesta Einsof:",
                          resultados['Einsof']['respuesta'].strip(),"---"]
            w_area = W.Textarea(
                value="\n".join(md_lineas),
                layout=W.Layout(width="100%",height="260px")
            )
            display(w_area)

            tabla_df.to_csv(ruta.with_suffix(".csv"))
            if any(v["limite_alcanzado"] for v in resultados.values()):
                print("Nota: Una respuesta alcanzó el límite máximo de tokens de generación.")

        except Exception:
            registro["error"] = traceback.format_exc()
            guardar()
            print(registro["error"], flush=True)
        finally:
            print("JSON guardado:", ruta)
            boton.disabled = False


boton.on_click(ejecutar)

display(W.VBox([
    W.HTML(value=(
        "<div style='background:linear-gradient(90deg,#0284c7,#0ea5e9);"
        "-webkit-background-clip:text;-webkit-text-fill-color:transparent;"
        "font-size:18px;font-weight:900;letter-spacing:2px;margin-bottom:4px;'>"
        "⚡ POLYDIM-EINSOF · Laboratorio A/B de Silicio</div>"
        "<div style='color:#64748b;font-size:11px;margin-bottom:8px;'>"
        "Stiefel FP64 · PMTP ShmBus · Multi-Agente · 0 tokens inter-agente</div>"
    )),
    alumno, token, guia_hf, modelo,
    W.HBox([caso, boton_azar]),
    prompt_preview,
    chk_custom, aviso_custom, txt_custom,
    cuadrado_einsof, boton, salida
]))
```
