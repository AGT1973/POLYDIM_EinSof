### Simulación de Silicio: Demostración del Vector Hand-Off sin Ruido Térmico

```python
# Simular estados latentes de la penúltima capa L-1 del Emisor
batch_size, seq_len, d_hidden = 1, 8, 2048
h_emisor = torch.randn(batch_size, seq_len, d_hidden, dtype=torch.float32)

# Extracción no destructiva de la secuencia de razonamiento (evitando mean-pooling K=8)
pooler = AttentionSummaryPooling(mode='last_token')
h_thought = pooler(h_emisor)
print(f'Tensor de pensamiento condensado sin pérdida de fase: shape={h_thought.shape}')

# Proyector de Impedancia Isométrico St(k, D) hacia Capa 0
projector = StiefelImpedanceProjector(d_hidden=2048, d_embed=1536)
embeds_adaptados = projector(h_emisor)
defect = projector.stiefel_defect()
print(f'Embeds adaptados para Capa 0: shape={embeds_adaptados.shape} | Defecto Stiefel: {defect:.2e}')

print('
[EXIT CODE 0] Kernel de hand-off verificado sin Domain Mismatch.')
```
